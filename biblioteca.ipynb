# Sistemas de gerenciamento de Livros

```python
import matplotlib.pyplot as plt   # pip install matplotlib

# PASSO 1 - Classe Livro
class Livro:
    def __init__(self, titulo, autor, genero, quantidade):
        self.titulo = titulo
        self.autor = autor
        self.genero = genero
        self.quantidade = quantidade

    def __str__(self):
        return (f"{self.titulo} por {self.autor} | Gênero: {self.genero} "
                f"| Disponíveis: {self.quantidade}")

# PASSO 2 - Lista de livros
biblioteca = []

# PASSO 3 - Funções
def cadastrar_livro(titulo, autor, genero, quantidade):
    biblioteca.append(Livro(titulo, autor, genero, quantidade))
    print(f"O livro '{titulo}' foi adicionado à biblioteca.")

def listar_livros():
    print("\nLivros na Biblioteca:")
    for livro in biblioteca:
        print(livro)

def buscar_livro(titulo):
    encontrados = [l for l in biblioteca if titulo.lower() in l.titulo.lower()]
    print(f"\nBuscando pelo título '{titulo}'...")
    print(*encontrados, sep="\n") if encontrados else print("Livro não encontrado na biblioteca.")
    return encontrados

# ------------------------------------------------------------
# PASSO 5 - Testar o sistema (cadastro e listagem)
# ------------------------------------------------------------
cadastrar_livro("Dom Quixote", "Miguel de Cervantes", "Romance", 3)
cadastrar_livro("Orgulho e Preconceito", "Jane Austen", "Romance", 2)
cadastrar_livro("1984", "George Orwell", "Ficção", 5)
cadastrar_livro("Cem Anos de Solidão", "Gabriel Garcia Marquez", "Realismo Mágico", 1)
cadastrar_livro("Apanhador no Campo de Centeio", "J.D. Salinger", "Ficção", 4)

listar_livros()          # Lista todos os livros
buscar_livro("1984")     # Busca existente
buscar_livro("Odisseia") # Busca inexistente


# ------------------------------------------------------------
# PASSO 4 - Gerar o gráfico de quantidade de livros por gênero
# ------------------------------------------------------------

# Dicionário (objeto do tipo mapping): chave = gênero, valor = total de exemplares.
quantidade_por_genero = {}
for livro in biblioteca:
    if livro.genero not in quantidade_por_genero:
        quantidade_por_genero[livro.genero] = 0        # Cria a chave na 1ª ocorrência
    quantidade_por_genero[livro.genero] += livro.quantidade  # Soma os exemplares

# Separa as chaves (gêneros) e os valores (quantidades) em duas listas.
generos = list(quantidade_por_genero.keys())
quantidades = list(quantidade_por_genero.values())

print("\nQuantidade de livros por gênero:", quantidade_por_genero)

# Cria um gráfico de barras
plt.bar(generos, quantidades, color='royalblue')

# Adiciona rótulos aos eixos
plt.xlabel('Gênero')
plt.ylabel('Quantidade de Livros')

# Adiciona um título ao gráfico
plt.title('Quantidade de Livros por Gênero')

# Adiciona o valor acima de cada barra (enumerate devolve posição e valor)
for i, valor in enumerate(quantidades):
    plt.text(i, valor, str(valor), ha='center', va='bottom')

plt.grid(axis='y')   # Linhas de grade horizontais para facilitar a leitura
plt.tight_layout()   # Ajusta o espaçamento do gráfico
plt.savefig('grafico_livros_por_genero.png')  # Salva a imagem do gráfico
plt.show()           # Mostra o gráfico
```
